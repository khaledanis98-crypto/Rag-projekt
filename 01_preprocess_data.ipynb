# 01 Preprocess data
Den här notebooken undersöker Netflix-datasetet och kontrollerar datakvalitet.

```
import pandas as pd

CSV_PATH = 'data/NetflixOriginals_clean.csv'
df = pd.read_csv(CSV_PATH)
df.head()
```

```
# Enkel datakvalitetskontroll

print("Completeness - saknade värden:")
print(df.isna().sum())

print("\nUniqueness - duplicerade rader:")
print(df.duplicated().sum())

print("\nValidity - kontroll av IMDB Score:")
print("Min:", df["IMDB Score"].min())
print("Max:", df["IMDB Score"].max())

print("\nConsistency - språk och genre:")
print("Antal språk:", df["Language"].nunique())
print("Antal genrer:", df["Genre"].nunique())
```

```
# Spara städad version igen
# Här är datan redan städad, men vi säkerställer format och tar bort eventuella dubletter.

df_clean = df.drop_duplicates().copy()
df_clean["Title"] = df_clean["Title"].astype(str).str.strip()
df_clean["Genre"] = df_clean["Genre"].astype(str).str.strip()
df_clean["Language"] = df_clean["Language"].astype(str).str.strip()
df_clean["Runtime"] = pd.to_numeric(df_clean["Runtime"], errors="coerce")
df_clean["IMDB Score"] = pd.to_numeric(df_clean["IMDB Score"], errors="coerce")

df_clean = df_clean.dropna(subset=["Title", "Genre", "Premiere", "Runtime", "IMDB Score", "Language"])
df_clean.to_csv("data/NetflixOriginals_clean.csv", index=False)

df_clean.shape
```
